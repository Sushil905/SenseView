# Dataset overview

Inspect the de-identified multi-view manifest and synthetic fixture. Keep subject-level splits intact; do not place direct identifiers here.

```python
from pathlib import Path
import pandas as pd

manifest = Path('../data/synthetic/manifest.csv')
if manifest.exists():
    df = pd.read_csv(manifest)
    display(df.head())
    print('rows:', len(df), 'sessions:', df.session_id.nunique(), 'subjects:', df.subject_id.nunique())
    display(df.groupby(['session_id', 'timestep']).view_id.nunique().describe())
    if 'expression' in df: display(df.expression.value_counts().sort_index())
else:
    print('Add a de-identified manifest before running this notebook.')
```
